## 2. Fields: feature / label / context / excluded

*Sort every field you plan to touch into these four buckets. Excluded needs a why.*

Features

gsc_impressions
gsc_clicks
gsc_avg_position
ga4_pageviews
sessions_organic
These variables are available before making a refresh decision.

Label / Proxy

Content Refresh Priority Score

No direct refresh label exists, so the priority score is treated as a proxy based on historical search performance.

Context

report_date
client_hash_id
content_hash_id
These fields identify the observation but are not prediction features.

Excluded

Future clicks, future impressions, future rankings and any information collected after the decision date are excluded because they would introduce data leakage.

```python
duckdb.sql("""
SELECT
    COUNT(*) AS total_rows,
    MIN(report_date) AS start_date,
    MAX(report_date) AS end_date
FROM page_daily
""").df()
```

## 3. Verify it with queries (grain, counts, missing values, windows)

*Every claim above gets a query cell here. A contract claim without a query next to it is a guess.*

Verification Queries

The following SQL queries verify the data contract by checking:

one row per content page per day
reporting window
data availability
missing values

```python
duckdb.sql("""
SELECT
    report_date,
    content_hash_id,
    COUNT(*) AS row_count
FROM page_daily
GROUP BY report_date, content_hash_id
ORDER BY row_count DESC
LIMIT 10
""").df()
```

## 4. Data limits

*What can this data never tell you? Unbalanced history, GSC-only early rows, window overlaps.*

Data Limits

This dataset contains historical search performance rather than direct evidence of why rankings changed.

The target is based on observed historical behaviour instead of a manually verified refresh label.

The sample represents only a subset of the full warehouse and therefore should be considered directional rather than complete.

Future information is intentionally excluded to avoid data leakage during model development.

```python
duckdb.sql("""
SELECT
    COUNT(*) AS total_rows,
    COUNT(*) FILTER (WHERE gsc_data_available IS TRUE) AS gsc_available,
    COUNT(*) FILTER (WHERE ga4_data_available IS TRUE) AS ga4_available
FROM page_daily
""").df()
```

```python
duckdb.sql("""
SELECT
    SUM(CASE WHEN gsc_impressions IS NULL THEN 1 ELSE 0 END) AS missing_impressions,
    SUM(CASE WHEN gsc_clicks IS NULL THEN 1 ELSE 0 END) AS missing_clicks,
    SUM(CASE WHEN gsc_avg_position IS NULL THEN 1 ELSE 0 END) AS missing_avg_position
FROM page_daily
""").df()
```
