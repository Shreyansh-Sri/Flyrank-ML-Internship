My Lane as an ML Task               

Chosen Lane: Content Refresh Prioritization

Task Type: Ranking / Scoring

The Content Refresh Prioritization lane is best framed as a ranking and scoring problem. The goal is to rank website pages according to how urgently they should be reviewed and refreshed. Instead of simply classifying pages into categories, the model assigns a priority score based on observed search performance signals such as trend direction, CTR, average search position, impressions, and content age. This helps content teams decide which pages should be updated first.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

Target or Proxy
The target is a content refresh priority score. Since there is no direct label indicating whether a page requires a refresh, a proxy target is created using observed search performance signals such as trend direction, click-through rate (CTR), average search position, impressions, and content age. Pages showing observed declining performance are assigned a higher refresh priority score based on the defined proxy.
